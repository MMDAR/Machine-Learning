```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(df['radius_mean'], df['texture_mean'], c=df['diagnosis'].map({'M': 1, 'B': 0}), cmap='coolwarm', alpha=0.7)
plt.xlabel('Radius Mean')
plt.ylabel('Texture Mean')
plt.title('Visualisasi 2D Data Breast Cancer')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df['radius_mean'], df['texture_mean'], df['area_mean'],
           c=df['diagnosis'].map({'M': 1, 'B': 0}), cmap='coolwarm', alpha=0.7)

ax.set_xlabel('Radius Mean')
ax.set_ylabel('Texture Mean')
ax.set_zlabel('Area Mean')
ax.set_title('Visualisasi 3D Data Breast Cancer')
plt.show()
```
