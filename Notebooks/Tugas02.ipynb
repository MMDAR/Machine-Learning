```python
import pandas as pd

df = pd.read_csv('../data/day.csv', sep=',')

# Acak data
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

n = len(df)
n_train = int(0.8 * n)
n_test = int(0.2 * n)
n_val = int(0.1 * n_train)

test = df[:n_test]
train_val = df[n_test:]

val = train_val[:n_val]
train = train_val[n_val:]

print("Jumlah Data Training:", len(train))
print("Jumlah Data Validation:", len(val))
print("Jumlah Data Testing:", len(test))

print("\n=== Data Training  ===")
print(train.head())

print("\n=== Data Validation  ===")
print(val.head())

print("\n=== Data ===")
print(test.head())
```
